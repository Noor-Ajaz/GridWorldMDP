```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Grid size
rows, cols = 5, 5

# Special states
goal_state = (4, 4)
negative_state = (3, 3)
obstacles = [(1, 1)]

# All valid states
states = [(r, c) for r in range(rows) for c in range(cols)
          if (r, c) not in obstacles]
```

```python
ACTIONS = ['U', 'D', 'L', 'R']
```

```python
def reward(state):
    if state == goal_state:
        return 10
    if state == negative_state:
        return -10
    return -0.1
```

```python
def move(state, action):
    r, c = state

    if action == 'U':
        r -= 1
    elif action == 'D':
        r += 1
    elif action == 'L':
        c -= 1
    elif action == 'R':
        c += 1

    next_state = (r, c)

    # Wall ya obstacle check
    if (r < 0 or r >= rows or
        c < 0 or c >= cols or
        next_state in obstacles):
        return state   # wapas wahi state

    return next_state

```

```python
def get_next_states(state, action):
    transitions = []

    intended = move(state, action)
    transitions.append((0.8, intended))

    other_actions = [a for a in ACTIONS if a != action]
    prob = 0.2 / len(other_actions)

    for a in other_actions:
        transitions.append((prob, move(state, a)))

    return transitions
```

```python
def value_iteration(gamma=0.9, theta=0.001, show_steps=False):
    V = {s: 0 for s in states}
    iteration = 0

    while True:
        delta = 0
        iteration += 1

        for s in states:
            if s in [goal_state, negative_state]:
                continue

            values = []
            for a in ACTIONS:
                total = 0
                for prob, s_next in get_next_states(s, a):
                    total += prob * (reward(s_next) + gamma * V[s_next])
                values.append(total)

            best_value = max(values)
            delta = max(delta, abs(V[s] - best_value))
            V[s] = best_value

        if show_steps:
            print(f"Iteration {iteration}:")
            print(V)
            print("-" * 40)

        if delta < theta:
            break

    return V
```

```python
def extract_policy(V, gamma=0.9):
    policy = {}

    for s in states:
        if s in [goal_state, negative_state]:
            policy[s] = 'T'
            continue

        best_action = None
        best_value = -1e9

        for a in ACTIONS:
            total = 0
            for prob, s_next in get_next_states(s, a):
                total += prob * (reward(s_next) + gamma * V[s_next])

            if total > best_value:
                best_value = total
                best_action = a

        policy[s] = best_action

    return policy
```

```python
def policy_iteration(gamma=0.9):
    policy = {s: np.random.choice(ACTIONS) for s in states}
    V = {s: 0 for s in states}

    stable = False
    iteration = 0

    while not stable:
        iteration += 1

        # Policy Evaluation
        for _ in range(20):
            for s in states:
                if s in [goal_state, negative_state]:
                    continue
                a = policy[s]
                total = 0
                for prob, s_next in get_next_states(s, a):
                    total += prob * (reward(s_next) + gamma * V[s_next])
                V[s] = total

        # Policy Improvement
        stable = True
        for s in states:
            if s in [goal_state, negative_state]:
                continue

            old_action = policy[s]
            best_action = extract_policy(V, gamma)[s]

            if best_action != old_action:
                policy[s] = best_action
                stable = False

        print(f"Policy Iteration Step {iteration} Complete")

    return policy, V
```

```python
def show_grid_values(V, title="Value Function"):
    grid = np.zeros((rows, cols))

    for (r, c), v in V.items():
        grid[r, c] = v

    plt.figure(figsize=(5,5))
    plt.imshow(grid, cmap='coolwarm')
    plt.colorbar()
    plt.title(title)
    plt.show()
```

```python
V = value_iteration(gamma=0.9, show_steps=True)
policy = extract_policy(V)

show_grid_values(V, "Value Iteration Result")

print("Optimal Policy:")
print(policy)
```

```python
policy_pi, V_pi = policy_iteration(gamma=0.9)
show_grid_values(V_pi, "Policy Iteration Result")

print("Policy Iteration Policy:")
print(policy_pi)
```
